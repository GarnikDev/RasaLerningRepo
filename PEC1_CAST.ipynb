Una empresa de bicicletas compartidas mantiene información de las estaciones y de las bicicletas disponibles.

```python
estaciones = ["BCN01", "BCN02", "BCN03", "BCN04", "BCN05"]
estaciones_activas = ["BCN01", "BCN03", "BCN04"]
bicicletas = [
    {"id": "B001", "estacion": "BCN01", "tipo": "electrica"},
    {"id": "B002", "estacion": "BCN02", "tipo": "mecanica"},
    {"id": "B003", "estacion": "BCN03", "tipo": "electrica"},
    {"id": "B004", "estacion": "BCN03", "tipo": "mecanica"},
    {"id": "B005", "estacion": "BCN04", "tipo": "electrica"},
    {"id": "B006", "estacion": "BCN05", "tipo": "mecanica"}
]
```

**Tareas:**

1. Obtén las estaciones que aparecen en los datos de bicicletas.
2. Obtén las estaciones que están activas y que además tienen alguna bicicleta registrada.
3. Mediante una comprensión de diccionario, crea un diccionario que tenga como clave el identificador de la bicicleta y como valor su tipo.
4. Crea una lista con los identificadores de las bicicletas eléctricas que pertenecen a estaciones activas.

SM **(2 puntos)**


````python
%load_ext pycodestyle_magic
````

````python
# Activamos las alertas de estilo
%pycodestyle_on
````

````python
# Escribe aquí tu solución.

estaciones: list[str] = ["BCN01", "BCN02", "BCN03", "BCN04", "BCN05"]
estaciones_activas: list[str] = ["BCN01", "BCN03", "BCN04"]
bicicletas: list[dict[str, str]] = [
    {"id": "B001", "estacion": "BCN01", "tipo": "electrica"},
    {"id": "B002", "estacion": "BCN02", "tipo": "mecanica"},
    {"id": "B003", "estacion": "BCN03", "tipo": "electrica"},
    {"id": "B004", "estacion": "BCN03", "tipo": "mecanica"},
    {"id": "B005", "estacion": "BCN04", "tipo": "electrica"},
    {"id": "B006", "estacion": "BCN05", "tipo": "mecanica"}
]
````

````python
# Tarea 1. Obtén las estaciones que aparecen en los datos de bicicletas.

# Utilizamos la propiedad de contener valores únicos de los conjuntos(set) y
# la comprensión de conjuntos para extraer todos los valores de la clave
# "estacion" de todas las bicicletas y guardarlos en un conjunto(set) llamado
# estaciones_en_bicicletas
estaciones_con_bicicletas: set[str] = {
    bicicleta["estacion"] for bicicleta in bicicletas
    }

# Tenemos en cuenta que puede ocurrir que ninguna de las estaciónes aparezcan
# en los datos de bicicletas, por lo que comprobamos si el conjunto(set) de
# estaciones_con_bicicletas está vacío, es decir, no hay estaciones con
# bicicletas. Así, si es el caso, en vez de mostrar un mensaje con una lista
# vacía, mostramos un mensaje indicando que no hay estaciones con bicicletas.
if (len(estaciones_con_bicicletas) == 0):
    print("No hay estaciones que aparezcan en los datos de bicicletas.")
else:
    # mostramos el resultado
    # (ordenamos las estaciones para que se vea más bonito)
    print("Las estaciones que aparecen en los datos de bicicletas son:",
          sorted(estaciones_con_bicicletas))
````

````python
# Tarea 2. Obtén las estaciones que no aparecen en los datos de bicicletas.

# Aprovechamos que ya tenemos un conjunto(set) con todas las estaciones que
# aparecen en los datos de bicicletas y una lista con todas las estaciones
# existentes y excluyendo las que aparecen de la lista de todas, obtenemos
# las que no aparecen.
estaciones_sin_bicicletas: set[str] = (
    set(estaciones) - estaciones_con_bicicletas
)
# Tenemos en cuenta que puede ocurrir que todas las estaciónes aparezcan en
# los datos de bicicletas, por lo que comprobamos si el conjunto(set) de
# estaciones_sin_bicicletas está vacío, es decir, no hay estaciones sin
# bicicletas. Así, si es el caso, en vez de mostrar un mensaje con una lista
# vacía, mostramos un mensaje indicando que no hay estaciones sin bicicletas.
if (len(estaciones_sin_bicicletas) == 0):
    print("No hay estaciones que no aparezcan en los datos de bicicletas.")
else:
    # mostramos el resultado
    # (ordenamos las estaciones para que se vea más bonito)
    print("Las estaciones que no aparecen en los datos de bicicletas son:",
          sorted(estaciones_sin_bicicletas))
````

````python
# 3. Mediante una comprensión de diccionario, crea un diccionario que tenga
# como clave el identificador de la bicicleta y como valor su tipo.

# Importamos esta librería para mostrar de
# manera más clara el diccionario resultante
import pprint

# Utilizamos la comprensión de diccionarios para generar nuevo diccionario
# comprimido {clave: valor for elemento in diccionario}.
# Seleccionando como clave el id de la bicicleta y como valor su tipo
bicicletas_por_tipo: dict[str, str] = {
    bicicleta["id"]: bicicleta["tipo"] for bicicleta in bicicletas
    }

# Utilizamos la librería pprint para mostrar
# de manera más clara el diccionario resultante
print("Diccionario con el id de la bicicleta como clave y su tipo como valor:")
pprint.pprint(bicicletas_por_tipo)
````

````python
# 4. Crea una lista con los identificadores de las bicicletas
# eléctricas que pertenecen a estaciones activas.

# Utilizamos la comprensión de listas
# [valor for elemento in diccionario if condicion]
# usando una condición(if) para extraer solo las bicicletas
# eléctricas y la estación de las cuales está incluida en la
# lista de estaciones activas, es decir, está en una estación activa.
# como resultado obtenemos una lista con los ids (bicicleta["id"])
# de las bicicletas con las condiciones del enunciado.
bicis_electric_activa: list[str] = [
    bicicleta["id"] for bicicleta in bicicletas
    if bicicleta["tipo"] == "electrica"
    and bicicleta["estacion"] in estaciones_activas]

# Mostramos el resultado
print(f"Ids de las bicicletas eléctricas que pertenecen a estaciones activas:\
\n{bicis_electric_activa}")
````

# Ejercicio 3

A partir de una lista de alquileres representados como diccionarios, se quiere construir un resumen por estación de origen.

```python
alquileres = [
    {"estacion": "BCN01", "tipo": "electrica", "duracion": 25},
    {"estacion": "BCN02", "tipo": "mecanica", "duracion": 18},
    {"estacion": "BCN01", "tipo": "mecanica", "duracion": 32},
    {"estacion": "BCN03", "tipo": "electrica", "duracion": 41},
    {"estacion": "BCN02", "tipo": "electrica", "duracion": 22},
    {"estacion": "BCN01", "tipo": "electrica", "duracion": 27}
]
```

**Tareas:**

1. Escribe una función `resumen_estaciones(alquileres)` que utilice `defaultdict` para agrupar las duraciones por estación.
2. La función debe devolver un diccionario donde cada estación tenga asociada la lista de duraciones de sus alquileres.
3. A partir del resultado, calcula el número de alquileres de cada estación.
4. Utiliza `OrderedDict` para devolver las estaciones ordenadas de mayor a menor número de alquileres.

SM **(2 puntos)**


````python
# Escribe aquí tu solución.

alquileres = [
    {"estacion": "BCN01", "tipo": "electrica", "duracion": 25},
    {"estacion": "BCN02", "tipo": "mecanica", "duracion": 18},
    {"estacion": "BCN01", "tipo": "mecanica", "duracion": 32},
    {"estacion": "BCN03", "tipo": "electrica", "duracion": 41},
    {"estacion": "BCN02", "tipo": "electrica", "duracion": 22},
    {"estacion": "BCN01", "tipo": "electrica", "duracion": 27}
]
````

````python
from collections import defaultdict

dur_por_est = defaultdict(list)

def resumen_estaciones(alquileres: list) -> dict:
    for alquiler in alquileres:
        estacion: str = alquiler['estacion']
        dur_por_est[estacion].append(alquiler['duracion'])
    return dur_por_est

resumen_est = resumen_estaciones(alquileres)
resumen_est
````

````python
alq_por_est: dict = {}
for est, list_dur in resumen_est.items():
    alq_por_est[est] = len(list_dur)

alq_por_est
````

````python
from collections import OrderedDict

ord_alq_por_est = OrderedDict(alq_por_est)
ord_alq_por_est
````

3. Obtén la duración expresada en minutos y segundos.
4. Utiliza `timedelta` para determinar la hora de finalización de un alquiler que comienza a las 18:45:00 y dura 75 minutos.

SM **(2 puntos)**


````python
# Escribe aquí tu solución.
import datetime

inicio: str = "2025-05-14 08:35:00"
fin:str = "2025-05-14 09:12:30"

inicio_dt: datetime = datetime.datetime.strptime(inicio, "%Y-%m-%d %H:%M:%S")
fin_dt: datetime = datetime.datetime.strptime(fin, "%Y-%m-%d %H:%M:%S")

duracion = fin_dt - inicio_dt
minutos: int = (int)(duracion.total_seconds() / 60)
segundos: int = (int)((duracion.total_seconds() / 60 - minutos) * 60)

print(f"La duracion es de {minutos} minutos y {segundos} segundos")
````

# Ejercicio 5

La empresa quiere representar cada estación mediante un objeto. Una estación tiene un identificador, una capacidad máxima y un número de bicicletas disponibles.

El identificador debe seguir el formato `EST-` seguido de dos letras mayúsculas y dos dígitos, por ejemplo `EST-BC01`.

**Tareas:**

1. Crea una clase `Estacion` con los atributos `identificador`, `capacidad` y `bicicletas_disponibles`.
2. En el constructor, valida mediante una expresión regular que el identificador tenga el formato correcto. Si no lo tiene, debe producirse una excepción apropiada.
3. Implementa un método `porcentaje_ocupacion()` que devuelva el porcentaje de ocupación de la estación.
4. Implementa un método `necesita_redistribucion()` que devuelva `True` cuando el porcentaje de ocupación sea inferior al 20 % o superior al 90 %.
5. Implementa un método `resumen()` que devuelva una cadena con la información principal de la estación.
6. Crea varios objetos `Estacion` y prueba todos los métodos.
7. Incluye al menos un identificador incorrecto y comprueba que la validación funciona correctamente.

SM **(2 puntos)**


````python
# Escribe aquí tu solución.
import re

class Estacion:
    def __init__(self, identificador: str, capacidad: int, bicicletas_disponibles: int) -> "Estacion":
        if re.fullmatch(r'EST-[A-Z]{2}\d{2}', identificador) != None:
            self.identificador = identificador
        else:
            raise Exception(f"El identificador debe seguir el formato \'EST- seguido\' de dos letras mayúsculas y dos dígitos, por ejemplo \'EST-BC01\'.")
        self.capacidad = capacidad
        self.bicicletas_disponibles = bicicletas_disponibles

    def porcentaje_ocupacion(self):
        return (self.bicicletas_disponibles) / (self.capacidad) * 100
    
    def necesita_redistribucion(self):
        return (self.porcentaje_ocupacion() < 20 or self.porcentaje_ocupacion() > 90)
    
    def resmen(self):
        mensaje = f"""identificador: {self.identificador}
capacidad: {self.capacidad}
bicicletas disponibles: {self.bicicletas_disponibles}
"""
        return mensaje
estacion1 = Estacion("EST-BC11", 200, 50)
estacion1.porcentaje_ocupacion()
print(estacion1.resmen())
````
